#### 3.5: G «Файловая статистика»

![3_5_G.png](photo/3_5_G.png)

```python
"""
test.txt
1 2 3 4 5
-5 -4 -3 -2 -1
10 20
20 10
"""
file_name = input()
with open(file_name, encoding="UTF-8") as file_in:
    numbers = [int(i) for i in file_in.read().split()]

print(len(numbers))
print(sum(1 for i in numbers if i > 0))
print(min(numbers))
print(max(numbers))
print(sum(numbers))
print(f"{sum(numbers) / len(numbers):.2f}")
```

#### 3.5: M «Обновление данных»

![3_5_M.png](photo/3_5_M.png)

```python
"""
test.json
{
  "one": "один",
  "three": "три",
  "two": "два"
}
"""
import json
from sys import stdin

file_name = input()
lines = []

with open(file_name, encoding="UTF-8") as file_in:
    records = json.load(file_in)
for line in stdin:
    lines.append(line.rstrip("\n").split(' == '))
for line in lines:
    records[line[0]] = line[1]
#Тут с GPT сделали потом изящнее, вместо двух циклов - один, я постоянно забываю о возможности передачи двух переменных
with open(file_name, 'w', encoding="UTF-8") as file_out:
    json.dump(records, file_out, ensure_ascii=False, indent=2)
```

```python
def number_length(number):
    result = len(str(abs(number)))
    return result
print(number_length(12345))
```

#### 4.1: D «Копейка рубль бережёт»

![4_1_D.png](photo/4_1_D.png)

```python
def take_small(money):
    result = [i for i in money if i < 100]
    return result
money = [1, 5, 200, 0.5, 0.05, 10, 25, 1000, 5000, 1, 2, 100, 0.1, 5, 2000, 0.01]
print(take_small(money))
```
