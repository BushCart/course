#### 2.2: C «Кто быстрее на этот раз?»

![2_2_C.png](photo/2_2_C.png)

```python
Petr = int(input())
Vasya = int(input())
Tolya = int(input())

tirdic = {
    Petr: 'Петя',
    Vasya: 'Вася',
    Tolya: 'Толя'
}

if Petr > Vasya:
    winner = Petr
else:
    winner = Vasya
if Tolya > winner:
    winner = Tolya
    
print(tirdic[winner])
```

#### 2.2: F «Сила прокрастинации»

![2_2_F.png](photo/2_2_F.png)

```python
year = int(input())

if year % 400 == 0:
    print('YES')
elif year % 100 == 0:
    print('NO')
elif year % 4 == 0:
    print('YES')
else:
    print('NO')
```

#### 2.3: E «Внимание! Акция!».

![2_3_E.png](photo/2_3_E.png)

```python
count = 0
while (numbers := float(input())) != 0:
    if numbers >= 500:
        numbers = numbers * 0.9
    count += numbers

print(count)
```
