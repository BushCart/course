#### 3.1: C «Анонс новости»

![3_1_C.png](photo/3_1_C.png)

```python
head_lenth = int(input())
count = int(input())
headline = [input() for i in range(count)]
for text in headline:
    if len(text) > head_lenth:
        print(f'{text[:head_lenth - 3]}...')
    else:
        print(f'{text}')
```

#### 3.1: M «Массовое возведение в степень»

![3_1_M.png](photo/3_1_M.png)

```python
count = int(input())
num_list = [int(input()) for i in range(count)]
exp = int(input())
for number in num_list:
    print(pow(number, exp))
```

#### 3.2: H «Кашееды — 4»

![3_2_H.png](photo/3_2_H.png)

```python
count = int(input())
child_dict = {}
for i in range(count):
    child, *porridge = input().split()
    for por in porridge:
        if por not in child_dict:
            child_dict[por] = []
        child_dict[por].append(child)
p = input()
if p in child_dict:
    print(*sorted(set(child_dict[p])), sep='\n') # Вот тут подсказал что вывод построчно, а не целиком
else:
    print('Таких нет')
```

#### 3.2: I «Зайка — 9»

![3_2_I.png](photo/3_2_I.png)

```python
words_dos = {}
words = input()
while words != "":
    for word in words.split():
        if word not in words_dos:
            words_dos[word] = 1
        else:
            words_dos[word] = words_dos.get(word) + 1
    words = input()

for key, value in words_dos.items():
    print(key, value)
```
